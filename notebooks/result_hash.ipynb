```python
##############################
#Horizon Scan Result Hash    #
#Maintainer: Christopher Chan#
#Date: 2024-09-08            #
#Version: 0.0.1              #
##############################

import os, sys, re
import pathlib
import hashlib
import pandas as pd

data_feature = pathlib.Path("../data/04_feature")
with open(f"{data_feature}/FINALwHM_r1_score.csv", "+r") as df:
    r1_score = pd.read_csv(df, sep = ",")

with open(f"{data_feature}/Wytham_Abbey_r2_tally.csv", "+r") as df:
    r2_score = pd.read_csv(df, sep = ",")

def sha256(value, length=6):
     sha256_hash = hashlib.sha256(value.encode()).hexdigest()
     return sha256_hash[:length]

print([i for i in r1_score.columns.values[5:] if not re.search(r"\d$", i)])
name_hashkey = {i: sha256(i) for i in r1_score.columns.values[5:] if not re.search(r"\d$", i)}
name_hashkey
```

```python
hashr1_score, hashr2_score = r1_score.copy(), r2_score.copy()

hashr1_score.columns = [name_hashkey[col] if col in name_hashkey else col for col in r1_score.columns]
hashr2_score.columns = [name_hashkey[col] if col in name_hashkey else col for col in r2_score.columns]
```

```python
# Hash outstanding name
hashr2_score.rename(columns = {"David McKay": sha256("David McKay")}, inplace = True)
hashr2_score.columns
```

```python
hashr1_score.to_csv(f"{data_feature}/hash_r1_score.csv", sep = ",")
hashr2_score.to_csv(f"{data_feature}/hash_r2_score.csv", sep = ",")
```
